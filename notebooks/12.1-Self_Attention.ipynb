# Checking a worked example from the lecture

The lecture works through a small causal (masked) attention example with $N=3$ tokens and $d_k=2$. On the slides the queries, keys, and values are given in **row form** (each row is a token), which is the convention used in most papers and in PyTorch:

$\mathbf{Q}=\mathbf{K}=\begin{bmatrix}1&0\\0&1\\1&1\end{bmatrix},\quad \mathbf{V}=\begin{bmatrix}1&0\\0&2\\3&1\end{bmatrix},\quad \mathbf{O}=\mathrm{Softmax}\left[\frac{\mathbf{Q}\mathbf{K}^{T}}{\sqrt{d_k}}\right]\mathbf{V}\ \text{(softmax over each row)}.$

The book stores tokens in the **columns** of a matrix instead, so every matrix is transposed and the products appear in the reverse order. Writing $\mathbf{Q}_c=\mathbf{Q}^{T}$, $\mathbf{K}_c=\mathbf{K}^{T}$, and $\mathbf{V}_c=\mathbf{V}^{T}$ for the column-form matrices, the book computes $\mathbf{O}_c=\mathbf{V}_c\,\mathrm{Softmax}\left[\frac{\mathbf{K}_c^{T}\mathbf{Q}_c}{\sqrt{d_k}}\right]$, with the softmax acting on each column, and the result is $\mathbf{O}_c=\mathbf{O}^{T}$. Both conventions compute exactly the same numbers.

The causal mask means that query $n$ may only attend to keys $m \leq n$. We implement it by setting the forbidden dot products to $-\infty$ before the softmax, so they receive zero attention. The cell below is self-contained (it does not use the functions you wrote above): it transposes the row-form matrices into columns, runs the book's column version, and compares the result with the row version.

```python
# Row form (rows are tokens), as on the lecture slides
Q_rows = np.array([[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]])
K_rows = Q_rows.copy()
V_rows = np.array([[1.0, 0.0], [0.0, 2.0], [3.0, 1.0]])
d_k = 2
n_tokens = Q_rows.shape[0]

# Column form (columns are tokens), as in the book
Q_cols = Q_rows.T
K_cols = K_rows.T
V_cols = V_rows.T

# Dot products: entry (m, n) is key m dotted with query n, scaled by sqrt(d_k)
scores_cols = K_cols.T @ Q_cols / np.sqrt(d_k)
# Causal mask: query n (column n) may only see keys m <= n (rows 0..n)
mask_cols = np.triu(np.ones((n_tokens, n_tokens)))
scores_cols = np.where(mask_cols == 1, scores_cols, -np.inf)
# Softmax over each column (exp(-inf) = 0, so masked keys get zero attention)
A_cols = np.exp(scores_cols) / np.sum(np.exp(scores_cols), axis=0, keepdims=True)
# Output: each column is a weighted sum of the value columns
O_cols = V_cols @ A_cols

# The same computation in row form (softmax over each row)
scores_rows = Q_rows @ K_rows.T / np.sqrt(d_k)
scores_rows = np.where(np.tril(np.ones((n_tokens, n_tokens))) == 1, scores_rows, -np.inf)
A_rows = np.exp(scores_rows) / np.sum(np.exp(scores_rows), axis=1, keepdims=True)
O_rows = A_rows @ V_rows

np.set_printoptions(precision=4, suppress=True)
print("Attention matrix, column form (column n = weights used by query n):")
print(A_cols)
print("Output, column form (column n = output for token n):")
print(O_cols)
print("\nTransposed back to row form:")
print("A =")
print(A_cols.T)
print("O =")
print(O_cols.T)
print("\nMatches the row-form computation:", np.allclose(A_cols.T, A_rows) and np.allclose(O_cols.T, O_rows))
print("Matches the lecture numbers:",
      np.allclose(A_cols.T, [[1, 0, 0], [0.3302, 0.6698, 0], [0.2483, 0.2483, 0.5035]], atol=1e-4) and
      np.allclose(O_cols.T, [[1, 0], [0.3302, 1.3395], [1.7587, 1.0000]], atol=1e-3))
np.set_printoptions()
```
